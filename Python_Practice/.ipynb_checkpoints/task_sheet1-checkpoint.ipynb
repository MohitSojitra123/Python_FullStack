## W.A.P. that replicates a calculatmor / Convert this to function/class based script when you learn about them

#### create naive script here

```python
a=int(input("Enter Value 1 : "))
b=int(input("Enter Value 1 : "))

print("Enter Operation sign \n 1 for + \n 2 for - \n 3 for * \n 4 for / \n 5 for % \n 6 for ** \n 7 for //")
opr=int(input("Enter Operation Sign Number: "))

if opr==1:
    print(f"{a} + {b} = {a+b}")
elif opr == 2:
    print(f"{a} - {b} = {a-b}")
elif opr == 3:
    print(f"{a} * {b} = {a*b}")
elif opr == 4:
    print(f"{a} / {b} = {a/b}")
elif opr == 5:
    print(f"{a} % {b} = {a%b}")
elif opr == 6 :
    print(f"{a} ** {b} = {a**b}")
elif opr == 7:
    print(f"{a} // {b} = {a//b}")
else:
    print("Enter Valid Sign : ")
    
```

## W.A.P. that calculates result of a student. Student has given exam for five subjects. Passing marks are 23 out of 70

#### create naive script here

```python
stu_name=input("Enter Student Name : ")
marks=[]
for i in range(0,5,1):
  marks.append(int(input(f"Enter Subject {i+1} Marks : ")))

for i in marks:
    if i < 23:
       print("Fail : ")
       break
else:
  percentage=(sum(marks)/(len(marks)*70))*100 
  print(f"Percentage : {round(percentage,2)}%") 
    
```

#### create function based script here

```python
def Marks_CalCulation():
    stu_name = input("Enter Student Name: ")

    marks = []

    for i in range(0, 5):
        marks.append(int(input(f"Enter Subject {i+1} Marks: ")))

    for i in marks:
        if i < 23:
            print("Fail!")
            break
    else:
        percentage = (sum(marks) / (len(marks) * 70)) * 100
        print(f"Student Name: {stu_name}")
        print(f"Percentage: {round(percentage, 2)}%")


Marks_CalCulation()
```

#### create class based script here

```python
class Result:
    def Marks_CalCulation(self):
        stu_name = input("Enter Student Name: ")
        marks = []
    
        for i in range(0, 5):
            marks.append(int(input(f"Enter Subject {i+1} Marks: ")))
    
        for i in marks:
            if i < 23:
                print("Fail!")
                break
        else:
            percentage = (sum(marks) / (len(marks) * 70)) * 100
            print(f"Student Name: {stu_name}")
            print(f"Percentage: {round(percentage, 2)}%")


student1=Result()
student1.Marks_CalCulation()
```

```python
a=int(input("Enter Number : "))

count=sum(range(1,a+1,1))
print(count)

while a>=1:
   j=1
   while j<=a:
       print(count,end=" ")
       count-=1
       j+=1
   print()    
   a-=1    
```
