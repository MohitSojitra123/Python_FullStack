## 2) Queue 

```python
Implement Queue
You script should contain three functions
1. enqueue
2. dequeue
3. peek
User will be given choice to perform action from above three.
Script should be kept running until user chooses to exit.
```

```python
list1=list()
size=int(input("Enter Queue Size : "))
current_size=1

def enqueue():
    global current_size
    global size
    if  current_size<=size:
        num=int(input(f"Enter Element : {current_size}/{size}"))
        list1.append(num)
        current_size+=1
    else:
        print("Queue Overflow : ")


def dequeue():
    global current_size
    global size
    if current_size==1:
        print("Queue in Underflow : ")
    else:
        ele=list1.pop(0)
        current_size-=1
        print(f"Dequeue Element : {ele}")
        

def show():
    print(list1)


def peek():
    if current_size==1:
        print("Queue Is Underflow")
    else:
        print(f"Peek Element :{list1[0]}")



while 1:
    print("Choose Option \n 1 For Enqueue \n 2 For Dequeue \n 3 For Show \n 4 For Peek \n 5 For Exit")
    num=int(input("Enter Number : "))

    if num == 1:
        enqueue()
    elif num==2:
        dequeue()
    elif num==3:
        show()
    elif num==4:
        peek()
    elif num==5:
        break
    
```

## 3) Circular Queue

```python
Implement Circular Queue
You script should contain four functions
1. Front: Get the front item from queue.
2. Rear: Get the last item from queue.
3. enqueue(value)
4. dequeue()
User will be given choice to perform action from above four.
Script should be kept running until user chooses to exit.
```

```python
list1=list()

size=int(input("Enter Size Of Circular Queue : "))
current_size=1


def FrontEnqueue():
    global current_size
    if current_size<=size:
        num=int(input(f"Enter Element (Front-Add) : {current_size}/{size}"))
        list1.insert(0,num)
        current_size+=1
    else:
        print("Circular Queue Is Overflow : ")


def RearEnqueue():
    global current_size
    if current_size<=size:
        num=int(input(f"Enter Element (Rear-Add) : {current_size}/{size}"))
        list1.append(num)
        current_size+=1
    else:
        print("Circular Queue Is Overflow : ")


def FrontDequeue():
    global current_size
    if current_size==1:
        print("Circular Queue Is Underflow : ")
    else:
        n=list1.pop(0)
        print(f"Front Dequeue : {n}")
        current_size-=1


def RearDequeue():
    global current_size
    if current_size==1:
        print("Circular Queue Is Underflow : ")
    else:
        n=list1.pop()
        print(f"Rear Dequeu : {n}")
        current_size-=1


def show():
    print(list1)

while 1:
    print("Choose Number \n 1 For FrontEnqueue \n 2 For RearEnqueue \n 3 For FrontDequeue \n 4 For RearDequeue \n 5 For Show \n 6 For Exit")
    user=int(input("Enter Option : "))

    if user==1:
        FrontEnqueue()
    elif user==2:
        RearEnqueue()
    elif user==3:
        FrontDequeue()
    elif user==4:
        RearDequeue()
    elif user==5:
        show()
    elif user==6:
        break 
    else:
        print("Enter Valid Option : ")
        
        
```

## 4) Selection Sort

```python
Selection sort
Use a static list
Use selection sort to sort this list
```

```python
def selection_sort(items):
    n = len(items)
    
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            if items[j] < items[min_idx]:
                min_idx = j
                
        if min_idx != i:
            items[i], items[min_idx] = items[min_idx], items[i]
            
    return items

my_list = [64, 34, 25, 12, 22, 11, 90]
sorted_list = selection_sort(my_list)
print("Sorted List:", sorted_list)
```

## 5) Bubble Sort

```python
Bubble sort
Use a static list
Use bubble sort to sort this list
```

```python
def bubble_sort(arr):
    n = len(arr)
    
    for i in range(n):
        swapped = False
        

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
                
        if not swapped:
            break

numbers = [64, 34, 25, 12, 22, 11, 90]
print("Original list:", numbers)

bubble_sort(numbers)
print("Sorted list:  ", numbers)
```

## 7) Binary Search

```python
Binary search
Use a static list
Use Binary search to find element asked by user
Use one of the sorting algorithm that you have developed to sort list here
```

```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2
        

        if arr[mid] == target:
            return mid
       
        elif arr[mid] < target:
            low = mid + 1
      
        else:
            high = mid - 1
            
   
    return -1


mylist = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(binary_search(mylist, 23))  
```

## 8) Matrix Multiplication

```python
Matrix Multiplication
1. Take two 3x3 matrix.
2. take user inputs to fill it up.
3. Multiply them both and store it in another 3x3 matrix.
4. Display the result.
```

```python

list1 = list()
list2 = list()

for i in range(1, 4):
    temp_list = list()

    for j in range(1, 4):
        num = int(input(f"Enter Number [{i}][{j}]: "))
        temp_list.append(num)

    list1.append(temp_list)

print("First Matrix:", list1)


for i in range(1, 4):
    temp_list = list()

    for j in range(1, 4):
        num = int(input(f"Enter Number [{i}][{j}]: "))
        temp_list.append(num)

    list2.append(temp_list)

print("Second Matrix:", list2)


ans = list()

for i in range(0, 3):
    temp_list = list()

    for j in range(0, 3):
        current_sum = 0

        for k in range(0, 3):
            current_sum += list1[i][k] * list2[k][j]

        temp_list.append(current_sum)

    ans.append(temp_list)

print("Multiplication:", ans)
```

## 9) Diterminent 

```python
Find diterminent of 3x3 matrix
```

```python
list1 = list()

for i in range(0, 3, 1):
    temp_list = list()
    for j in range(0, 3, 1):
        num = int(input("Enter Number : "))
        temp_list.append(num)
    list1.append(temp_list)

print(list1)

determinant = (
    (list1[0][0] * ((list1[1][1] * list1[2][2]) - (list1[2][1] * list1[1][2])))
    - (list1[0][1] * ((list1[1][0] * list1[2][2]) - (list1[2][0] * list1[1][2])))
    + (list1[0][2] * ((list1[1][0] * list1[2][1]) - (list1[2][0] * list1[1][1])))
)

print(f"Determinant : {determinant}")
```

## 11) Random Game

```python
Randomly generate a 4-digit number. Ask the user to guess a 4-digit number. For every digit that the user
guessed correctly in the correct place, they have a “cow”. For every digit the user guessed correctly in the
wrong place is a “bull.” Every time the user makes a guess, tell them how many “cows” and “bulls” they
have. Once the user guesses the correct number, the game is over. Keep track of the number of guesses
the user makes throughout teh game and tell the user at the end.

Say the number generated by the computer is 1038. An example interaction could look like this:
 Welcome to the Cows and Bulls Game!
 Enter a number:
 >>> 1234
 2 cows, 0 bulls
 >>> 1356
 1 cow, 1 bull
 ...
Until the user guesses the number.
```

```python
import random

ran=str(random.randint(1000,9999))
list1=list()

print(ran)

while 1:
    num=input("Enter Guess Number : ")
    
    if num==ran:
        print("Correct : ")
        print(list1)
        break
    else:
        for i in range(0,4,1):
            if(ran[i]==num[i]):
                print("Cow",end="-")
            else:
                print("Bull",end="-")
        print()
        list1.append(num)
```

## 15)  LCM - GCM

```python
Find LCM and GCM of a number
```

```python

a = int(input("Enter Value 1 : "))
b = int(input("Enter Value 2 : "))

big = 0
small = 0


if a > b:
    big = a
else:
    big = b

# LCM
while 1:
    if big % a == 0 and big % b == 0:
        print(f"LCM : {big}")
        break
    else:
        big += 1

if a > b:
    small = b
else:
    small = a


# GCD
for i in range(1, small + 1):
    if a % i == 0 and b % i == 0:
        gcd = i

print(f"GCD : {gcd}")
```
