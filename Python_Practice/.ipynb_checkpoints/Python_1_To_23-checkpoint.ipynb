# break, continue, pass

- __break:__ Break the loop and exits the code block.
- __continue:__ Skips the next part of loop and __continues__ to next iteration.
- __pass:__ Does nothing.

## Pass

````python
def f(arg):
    pass    

class C: 
    pass       

for i in range(10):
    if i%2 == 0:
        print(i, end=' ')
    else:
        pass 
````

## Break

````python
a_list = list(range(100))

for num in a_list:
    if num > 10:
        break
    print(num, end=' ')
````

## Continue

````python
a_list = list(range(100))
for num in a_list:
    if num%5 != 0:
        continue
    print(num, end=' ')
````

# else with loops

- Loop statements may have an else clause.
- It is executed when the loop terminates through exhaustion of the list with __for__.
- When the condition becomes false with __while__.
- But __not__ when the loop is __terminated by a break statement__

## For 

````python
for num in range(10):
    if num > 11:
        break
    print(num, end=' ')
else:
    print("The loop was executed completely")
````

````python
is_prime = 37
for num in range(2, is_prime//2):
    if is_prime%num == 0:
        print(f"{is_prime} has a factor {num} * {is_prime/num}")
        break
else:
    print(f"{is_prime} is a prime number")
````

## While

````python
is_prime = 37
num = 2
limit = is_prime//2

while num<limit:

    if is_prime%num == 0:
        print(f"{is_prime} has a factor {num} * {is_prime/num}")
        
        break
    num += 1
    
else:
    print(f"{is_prime} is a prime number")
````

## Recursion

````python
def recur_fibo(n):
    
    if n <= 1:
        return n
    
    else:
        return(recur_fibo(n-1) + recur_fibo(n-2))

nterms = 10


if nterms <= 0:
    print("Plese enter a positive integer")

else:
    print("Fibonacci sequence:")
    
    for i in range(nterms):
        print(recur_fibo(i))
````

## Slicing (Lab-8)

- For simplicity we will be using lists in examples.
- __a\[ : \]:__ Means every object in a.
- __a\[ i : \]:__ Means start __from i<sup>th</sup> index__ and cut __till the last index.__ (including last index)
- __a\[ : i \]:__ Means start __from 0<sup>th</sup> index__ and cut __till i<sup>th</sup> index__ (not including i<sup>th</sup> index).
- __a\[ i : j \]:__ Means start __from i<sup>th</sup> index__ and cut __till j<sup>th</sup> index__ (not including j<sup>th</sup> index).
- __a\[ i : j : k \]:__ Means start __from i<sup>th</sup> index__ take __k steps each time__ and cut __till j<sup>th</sup> index__ (not including j<sup>th</sup> index) __default k = 1.__
- __a\[ : : k \]:__ Means if __k is positive__ then start __from 0<sup>th</sup> index__ and cut __till last index__ (including last index) taking __k steps each time__.

````python
a_list = list(range(1, 11))
````

````python
print(a_list[:])
````

````python
b_list = a_list[:]
print(b_list)
````

````python
del a_list[:]  
print(a_list)  

del a_list  
print(a_list) 
````

## Negative Index and Negative steps

- __a\[ -i : \]:__ Means start __from -i<sup>th</sup> last index__ and cut __till the last index.__ (including last index)
- __a\[ : -i \]:__ Means start __from 0<sup>th</sup> index__ and cut __till i<sup>th</sup> last index__ (not including i<sup>th</sup> last index).
- __a\[ -i : -j \]:__ Means start __from -i<sup>th</sup> last index__ and cut __till -j<sup>th</sup> last index__ (not including -j<sup>th</sup> last index). __-i < -j__
- __a\[ -i : -j : -k \]:__ Means start __from i<sup>th</sup> last index__ take __k steps in reverse each time__ and cut __till j<sup>th</sup> last index__ (not including j<sup>th</sup> last index) __default k = 1.__ __-i > -j__
- __a\[ : : -k \]:__ Means if __k is negative__ then start __from -1<sup>th</sup> index__ and cut __till first index__ (including first index) taking __k steps in reverse each time__.

````python
class Team:
    '''
    Contains List of Junior team members and also overrides the __iter__() function.
    '''
    def __init__(self):
        self._junior_members = list()
  
    def add_junior_members(self, members):
        self._junior_members += members
  
    def __iter__(self):
        ''' Returns the Iterator object '''
        return TeamIterator(self)

class TeamIterator:
    ''' Iterator class '''
    def __init__(self, team):
        self._team = team
        self._index = 0
 
    def __next__(self):
        ''''Returns the next value from team object's lists '''
        if self._index < len(self._team._junior_members) :
            
            result = self._team._junior_members[self._index]
            
            self._index += 1
            return result
        
        raise StopIteration
    def __iter__(self):
        return self
````

````python
team = Team()

team.add_junior_members(['Sam', 'John', 'Marshal'])

for member in team:
    print(member)
    
print("Sam" in team)
````

````python
class Team:
    '''
    Contains List of Junior and senior team members and also overrides the __iter__() function.
    '''
    def __init__(self):
        self._junior_members = list()
  
    def add_junior_members(self, members):
        self._junior_members += members
  
    def __iter__(self):
        ''' Returns the Iterator object '''
        return iter(self._junior_members)
    
    def __next__(self):
        return next(self)
````

````python
team = Team()

team.add_junior_members(['Sam', 'John', 'Marshal'])

for member in team:
    print(member)
````

````python
from itertools import *  
numbers = [1, 2, 3]
characters = ['a', 'b', 'c']


chained_iterables = chain(numbers, characters)

print(f"TYPE: {type(chained_iterables)}")

print(f"Calling next: {next(chained_iterables)}")
print(f"Calling next: {next(chained_iterables)}")
print(f"Calling next: {next(chained_iterables)}")
print(f"Calling next: {next(chained_iterables)}")
print(f"Calling next: {next(chained_iterables)}")
````

````python
""" 

let's say we have a business logic and that produces the iterables that we do not know in advance
below genrator creates two lists 
one with odd numbers less than 50 and 
another with even numbers greater or equal to 50

"""

def make_iterables_to_chain():
    
    num_odd_lt_50 = []
    num_even_gt_50 = []
    
    for num in range(100):
        if num%2 and num < 50:
            num_odd_lt_50.append(num)
        elif num >= 50 and not num%2:
            num_even_gt_50.append(num)
    
    yield num_odd_lt_50
    
    yield num_even_gt_50
````

````python

chained_iterables = chain.from_iterable(make_iterables_to_chain())

print(f"TYPE: {type(chained_iterables)}")

print(f"Calling next: {next(chained_iterables)}")
print(f"Calling next: {next(chained_iterables)}")
print(f"Calling next: {next(chained_iterables)}")
print(f"Calling next: {next(chained_iterables)}")
````

````python
for i in chain.from_iterable(make_iterables_to_chain()):
    print(i, end=' ')
````

## Zip Longest

- It uses __None__ as default to fill value of shorter iterable when it is exhausted.

````python
def i_have_defaults(first, second=2, third=3, zero=0):  
    
    print(f"\nfirst: {first}")
    print(f"\nsecond: {second}")
    print(f"\nthird: {third}")
    print(f"\nzero: {zero}")
    
    if zero == 0:  
        print("My value was not changed :|")
        
    elif zero > 0:
        print("I am bigger now :D")
    
    else:
        print("You made me smaller :(")
        
i_have_defaults(first=1)
i_have_defaults(1, zero=1)
i_have_defaults(1, zero=-1)
````

## \*args and \*\*kwargs

````python
def variable_length_args(*args):
    
    print(f"type of args: {type(args)}")
    
    for arg in args:
        print(arg)
        
variable_length_args('a', 1, (2, 3,), [2, 'b'], 2.14)
````

````python
def variable_length_kwargs(**kwargs):
    
    print(f"type of kwargs: {type(kwargs)}")
    
    for key, value in kwargs.items():
        print(f"{key} = {value}")
        
variable_length_kwargs(first='a', second=1, third=(2, 3,), fourth=[2, 'b'], fifth=2.14)
````

````python
class DeepChainMap(collections.ChainMap):
    'Variant of ChainMap that allows direct updates to inner scopes'

    def __setitem__(self, key, value):
        for mapping in self.maps:
            if key in mapping:
                mapping[key] = value
                return
        self.maps[0][key] = value

    def __delitem__(self, key):
        for mapping in self.maps:
            if key in mapping:
                del mapping[key]
                return
        raise KeyError(key)
        

d = DeepChainMap({'zebra': 'black'}, {'elephant': 'blue'}, {'lion': 'yellow'})

d['lion'] = 'orange' 

d['snake'] = 'red'   

del d['elephant']   

print(d.items())

````

## Class and Instance Variables

- __Instance variables__ are for data __unique to each instance__. 
- __Class variables__ are for attributes and methods __shared by all instances__ of the class.

## Inheritance

````python
class A:
    def method_a(self):
        print(f"memory addr of obj in A: {id(self)}")
        print("method_a from A")
        
    def method_b(self):
        self.method_a()
        
class B(A):
    def method_a(self):
        print(f"memory addr of obj in B: {id(self)}")
        print("method_a from B")
#         A.method_a(self)
        super().method_a()
    
````

````python
b = B()
b.method_b()
````

````python
class Mapping:
    def __init__(self, iterable):
        self.items_list = []
        self.__update(iterable)

    def update(self, iterable):
        for item in iterable:
            self.items_list.append(item)

    __update = update   # private copy of original update() method

class MappingSubclass(Mapping):

    def update(self, keys, values):
        # provides new signature for update()
        # but does not break __init__()
        for item in zip(keys, values):
            self.items_list.append(item)
````

````python
class Employee:
    
    num_of_emps = 0          # Always same for all the instances of class
    raise_amount = 1.04      # May differ from employee to employee
    
    def __init__(self, first_name, last_name, salary):
        self.first_name = first_name
        self.last_name = last_name
        self.salary = salary
        self.email = first_name + '.' + last_name + "@inexture.com"
        
        Employee.num_of_emps +=  1
        
        
    def full_name(self):
        return f"{self.first_name} {self.last_name}"
    
    def apply_raise(self):
        print(f"Raising the salary with amount: {self.raise_amount}")
        self.salary = int(self.salary*self.raise_amount)
        
    @classmethod
    def set_raise_amount(cls, amount):
        cls.raise_amount = amount
    
    @classmethod
    def from_string(cls, string):
        first_name, last_name, salary = string.split('-')
        return cls(first_name, last_name, int(salary))
````

````python
class Employee:
    
    num_of_emps = 0          # Always same for all the instances of class
    raise_amount = 1.04      # May differ from employee to employee
    
    def __init__(self, first_name, last_name, salary):
        self.first_name = first_name
        self.last_name = last_name
        self.salary = salary
        self.email = first_name + '.' + last_name + "@inexture.com"
        
        Employee.num_of_emps +=  1
        
        
    def full_name(self):
        return f"{self.first_name} {self.last_name}"
    
    def apply_raise(self):
        print(f"Raising the salary with amount: {self.raise_amount}")
        self.salary = int(self.salary*self.raise_amount)
        
    @classmethod
    def set_raise_amount(cls, amount):
        cls.raise_amount = amount
    
    @classmethod
    def from_string(cls, string):
        first_name, last_name, salary = string.split('-')
        return cls(first_name, last_name, int(salary))
    
    @staticmethod
    def is_weekend(day):
        return day.weekday() == 5 or day.weekday() == 6
````

## decorator (Lab-20)

## Example Basic

````python
# 1. Define the decorator function
def my_decorator(func):
    def wrapper():
        print("Something is happening BEFORE the function runs.")
        func()  # Executes the original function
        print("Something is happening AFTER the function runs.")
    return wrapper

# 2. Apply the decorator using the @ symbol
@my_decorator
def say_hello():
    print("Hello, World!")

# 3. Call the function
say_hello()
````

````python
def null_decorator(func):
    return func
def greet():
    return 'Hello!'

greet_decorated = null_decorator(greet)

greet_decorated()
````

````python
@null_decorator
def greet():
    return 'Hello!'

greet()
````

## Multiple Decorators

````python
def strong(func):
    def wrapper():
        return '<strong>' + func() + '</strong>'
    return wrapper

def emphasis(func):
    def wrapper():
        return '<em>' + func() + '</em>'
    return wrapper
````

````python
@emphasis
@strong
def greet():
    return 'Hello!'
````

````python
greet()
````

````python
def trace(func):
    def wrapper(*args, **kwargs):
        print(f'TRACE: calling {func.__name__}() '
              f'with args: {args}, kwargs: {kwargs}')

        original_result = func(*args, **kwargs)

        print(f'TRACE: {func.__name__}() '
              f'returned {original_result!r}')

        return original_result
    return wrapper
````

````python
@trace
def say(name, line):
    return f'{name}: {line}'
````

````python
say('Jane', 'Hello, World')
````

````python
import functools

def uppercase(func):
    @functools.wraps(func)
    def wrapper():
        return func().upper()
    return wrapper
````

````python
@uppercase
def greet():
    """Return a friendly greeting."""
    return 'Hello!'
````

````python
x = "a"
def outer():
    x = "b"
    def inner():
        x = "c"
        print("from inner:", x)

    inner()
    print("from outer:", x)

outer()
print("globally:", x)
````

````python
x = "a"
def outer():
    x = "b"
    def inner():
        nonlocal x
        x = "c"
        print("inner:", x)

    inner()
    print("outer:", x)

outer()
print("global:", x)
````

````python
x = "a"
def outer():
    x = "b"
    def inner():
        global x
        x = "c"
        print("inner:", x)

    inner()
    print("outer:", x)

outer()
print("global:", x)
````

````python
def outer_function():
    x = "original outer"
    
    def inner_function():
        nonlocal x
        x = "changed outer"
        
    inner_function()
    print(x)  # Outputs: changed outer

outer_function()
````


## 3. Main Difference: Iterable vs itertools

| Iterable | itertools |
|---|---|
| An object that can be looped over | A Python module |
| Contains or produces elements | Provides tools to process iterables |
| Example: List, Tuple, String | Example: `product()`, `chain()`, `islice()` |
| Used with `for` loops | Used to perform advanced iteration operations |

## 5. Important itertools Functions

| Function | Simple Meaning |
|---|---|
| `count()` | Generate numbers continuously |
| `cycle()` | Repeat iterable elements |
| `repeat()` | Repeat the same value |
| `chain()` | Combine multiple iterables |
| `islice()` | Extract selected elements |
| `tee()` | Create multiple independent iterators |
| `product()` | Generate Cartesian products |
| `permutations()` | Generate different arrangements |
| `combinations()` | Generate selections |


| Concept | Meaning |
|---|---|
| `return` | Return a value and terminate |
| `yield` | Return a value and pause |
| Generator | Function containing `yield` |
| `next()` | Get the next generated value |
| `StopIteration` | Indicates that the generator is exhausted |

### Remember:

- **return = Give the result and stop.**
- **yield = Give one value, pause, and continue later.**
